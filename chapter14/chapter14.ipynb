```python
def greet():
    print("안녕하세요!")
    print("파이썬 공부를 시작합니다.")

greet()
greet()
```

```python
# 1. 함수에 파라미터(매개변수) 추가
# 2. 함수 안에서 파라미터 사용
# 3. 함수 호출시 파라미터 넘겨주기

def greet(name, study):
    print(f"안녕하세요! {name}님")
    print(f"{study} 공부를 시작합니다.")

greet("홍길동", "SQL")
greet("김철수", "Python")
```

```python
# 두 개의 숫자를 입력 받아서 더한 값을 반환하는 함수
# 함수 안에 print하면 지저분해지기 때문에 return하는게 일반적이다.

def first_num(num1, num2):
    return num1 + num2

num1 = int(input("첫 번째 숫자 : "))
num2 = int(input("두 번째 숫자 : "))

result = first_num(num1,num2)

print("두 수의 합은:", result)
```

```python
# 두 수(0~9까지)를 입력받아서 사칙연산 결과를 반환하는 calc 라는 함수를 만드시오
# 4개 결과값을 한번에 반환(Tuple)
# 사용자가 입력한 값이 숫자가 아닐 경우 오류메시지 출력하고 프로그램 종료

def calc(a, b):
    num_1 = a + b
    num_2 = a - b
    num_3 = a * b
    num_4 = a / b
    return num_1, num_2, num_3, num_4

a = int(input("첫 번째 숫자 : "))
b = int(input("두 번째 숫자 : "))

numbers = (calc(a,b))

print(numbers)
print(f"첫 번째 숫자는 {a}입니다.")
print(f"두 번째 숫자는 {b}입니다.")
```

```python
# 두 수(1~9까지)를 입력받아서 사칙연산 결과를 반환하는 calc 라는 함수를 만드시오
# 4개 결과값을 한번에 반환(Tuple)
# 사용자가 입력한 값이 숫자가 아닐 경우 오류메시지 출력하고 프로그램 종료

def calc(num1, num2):
    sum = num1 + num2
    minus = num1 - num2
    multi = num1 * num2
    div = num1 / num2
    return sum, minus, multi, div

first = input("첫 번째 숫자 : ")
second = input("두 번째 숫자 : ")

lenth = len(first) + len(second)

if (lenth == 2 and ord(first) >= 49 and ord(first) <= 57) and (lenth == 2 and ord(second) >= 49 and ord(second) <= 57):
    result = calc(int(first), int(second))
    print(type(result))
else:
    print("숫자가 아닙니다. 입력 값을 확인하세요.")
```

< 데일리 퀴즈 보충 >

| 구분      | 방식             | 예시                      |
| ------- | -------------- | ----------------------- |
| **1세대** | `%` 연산자        | `"이름: %s" % name`       |
| **2세대** | `str.format()` | `"이름: {}".format(name)` |
| **3세대** | `f-string`     | `f"이름: {name}"`         |

```python
# 1세대 - (% formatting) % 연산자 방식 (%s, %d처럼 자료형에 맞는 서식 지정자 사용)

# %s → 문자열
# %d → 정수
# %f → 실수

name = "민수"
age = 20

print("이름: %s, 나이: %d" % (name, age))

# --------------------------------------------
# 2세대 - str.format() 방식 ({} 자리에 format()에 전달한 값이 순서대로 들어감)
# {} ← name
# {} ← age

print("이름: {}, 나이: {}".format(name, age))

# --------------------------------------------
# 3세대 - f-string 방식 (문자열 앞에 f를 붙이고 {} 안에 변수나 표현식을 직접 작성 가능) // 가독성이 높다

score = 85

print(f"{score=}")
```
