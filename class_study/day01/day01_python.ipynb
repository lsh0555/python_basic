## MISSION 01

* 해결하고 싶은 문제 : 여러 숫자의 합계를 계산한다.
* 입력 : 여러 숫자들
* 처리 : 여러 숫자들을 더한다.
* 출력 : 숫자들의 더한 값
* AI에게 도움받고 싶은 부분 : 파이썬 코드로 실행하고 싶은데 코드 짜는게 어려워

Q. ME
나는 파이썬을 처음 배우고 있어.
아래 반복 작업을 나중에 파이썬으로 해결해 보고 싶어.
[ 해결하고 싶은 문제 : 여러 숫자의 합계를 계산한다.
입력 : 여러 숫자들
처리 : 여러 숫자들을 더한다.
출력 : 숫자들의 더한 값 ]
아직 코드는 작성하지 말고,
내가 문제를 더 명확하게 정의하기 위해 생각해야 할 질문을 3개만 만들어 줘.

A. GPT
1. 입력되는 숫자의 개수는 항상 정해져 있나요, 아니면 매번 달라질 수 있나요?
2. 입력되는 숫자는 정수만 있나요, 아니면 소수도 포함될 수 있나요?
3. 합계를 계산한 뒤에는 최종 합계만 출력하면 되나요, 아니면 입력한 숫자들도 함께 보여줘야 하나요?

Q. ME
price = 12000

quantity = 3

total = price * quantity

print(total)

수량을 0으로 바꾸면 결과는 어떻게 될까?
가격을 15000으로 바꾸면 어떤 값이 출력될까?
price * quantity를 직접 계산한 값과 프로그램 결과가 같은가?

A. GPT
현재 출력값은 36,000이야.

수량을 0으로 바꾸면?
12000 × 0 = 0이므로 0이 출력돼.
가격을 15000으로 바꾸면?
15000 × 3 = 45000이므로 45000이 출력돼.
직접 계산한 값과 프로그램 결과가 같은가?
12000 × 3 = 36000이고 프로그램도 36000을 출력하므로 같아.
