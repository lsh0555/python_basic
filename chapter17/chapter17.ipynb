```python
# 예상 가능한 예외 처리 : try-except
# 모든 오류를 자동으로 해결하는 문법이 아님, 
# 문제 생길 모든 코드에 try-except 사용하면 해결 해야할 코드도 다 넘어간다.. 적절하게 사용해야함
# try -> 문제가 발생할 가능성이 있는 코드 / except -> 그 문제에 대한 대응


try:
    age = int(input("나이를 입력하세요: "))
    print(f"입력한 나이: {age}")

except ValueError:
    print("나이는 숫자로 입력해 주세요.")
```

```python
try:
    number = int(input("숫자를 입력하세요: "))     # ValueError와 ZeroDivisionError 발생 가능성
    result = 100 / number
    print(result)

except ValueError:
    print("정수를 입력해 주세요.")

except ZeroDivisionError:
    print("0으로 나눌 수 없습니다.")
```

```python
# try-except-else-finally


try:   # -> → 예외가 발생할 수 있는 작업 시도
    number = int(input("정수: "))

except ValueError:   # → 지정한 예외가 발생했을 때 대응
    print("정수가 아닙니다.")

else:   # → try에서 예외가 없었을 때 실행
    print(f"입력 성공: {number}")

finally:   # → 성공·실패와 관계없이 마지막에 실행 // 성공 실패 상관없이 무조건 실행해야할 때 사용
    print("입력 처리를 마쳤습니다.")
```

```python
try:
    number = int(input("정수: "))
    print(f"입력 성공: {number}")   # 에러가 나면 이 부분은 실행 X

except ValueError:
    print("정수가 아닙니다.")

finally:
    print("입력 처리를 마쳤습니다.")
```

```python
while True:

    try:
        score = int(input("점수(0~100): "))

    except ValueError:
        print("숫자로 입력해 주세요.")
        continue

    if 0 <= score <= 100:
        break

    print("0부터 100 사이의 값을 입력해 주세요.")

print(f"입력된 점수: {score}")
```
